```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import emcee
import corner

# ==========================================================
# COSTANTI FISICHE
# ==========================================================
c = 2.99792458e10
G = 6.67430e-8
h = 6.62607015e-27
kB = 1.380649e-16
sigma = 5.670374419e-5
Msun = 1.98847e33

# ==========================================================
# FILE INPUT
# ==========================================================
file_linee = "Tab_completa_revisione.csv"
file_transizioni = "dati.asc"
```

```python
# ==========================================================
# PARAMETRI SS
# ==========================================================
eta = 0.083
Nr = 80
Nnu = 60
nu_min = 1e13
nu_max = 1e17

# ==========================================================
# PARAMETRI MCMC
# ==========================================================
N_WALKERS = 32
N_STEPS = 5000
BURN_IN = 1000
THIN = 10

CF_BLR = 0.1
REL_ERR = 0.20

# prior richieste da te
LOGM_MIN, LOGM_MAX = 7.0, 10.0
LOGMDOT_MIN, LOGMDOT_MAX = -2.0, 0.0
```

```python
# ==========================================================
# FUNZIONI UTILI
# ==========================================================
def trova_colonna(df, possibili_nomi, obbligatoria=True):
    for nome in possibili_nomi:
        if nome in df.columns:
            return nome
    if obbligatoria:
        raise ValueError(f"Non trovo nessuna di queste colonne: {possibili_nomi}")
    return None


def leggi_transizioni(file_transizioni):
    names, Eion, nu_ion_list = np.genfromtxt(
        file_transizioni,
        dtype=("U50", float, float),
        unpack=True
    )

    return pd.DataFrame({
        "line_name": names,
        "E_ion": Eion,
        "nu_ion": nu_ion_list
    })


def estrai_linee_osservate(row, norm_columns):
    risultati = []

    for col in norm_columns:
        valore = row[col]
        if pd.isna(valore) or valore <= 0:
            continue

        line_name = col.replace("_norm", "")
        unc_col = f"{line_name}_norm_unc"

        L_line = valore * 1e42

        if unc_col in row.index and pd.notna(row[unc_col]) and row[unc_col] > 0:
            sigma_L = row[unc_col] * 1e42
        else:
            sigma_L = np.nan

        risultati.append({
            "line_name": line_name,
            "norm_value": valore,
            "L_line_obs": L_line,
            "sigma_L": sigma_L
        })

    return pd.DataFrame(risultati)
```

```python
# ==========================================================
# SPETTRO SS
# ==========================================================
def costruisci_spettro_ss(
    MBH_solar,
    mdot_edd,
    eta=0.083,
    Nr=400,
    Nnu=120,
    nu_min=1e13,
    nu_max=1e17
):
    MBH = MBH_solar * Msun
    Ledd = 1.3e38 * MBH_solar
    Ldisk = mdot_edd * Ledd
    Mdot = Ldisk / (eta * c**2)

    Rs = 2 * G * MBH / c**2
    Rin = 3 * Rs
    Rout = 3000 * Rs
    R = np.logspace(np.log10(Rin), np.log10(Rout), Nr)

    f = 1.0 - np.sqrt(Rin / R)
    f = np.clip(f, 0.0, None)

    T = ((3 * G * MBH * Mdot) / (8 * np.pi * sigma * R**3) * f) ** 0.25

    nu = np.logspace(np.log10(nu_min), np.log10(nu_max), Nnu)

    Bnu = np.zeros((Nnu, Nr))
    for j in range(Nr):
        if T[j] <= 0:
            continue
        x = h * nu / (kB * T[j])
        x = np.clip(x, 1e-12, 700.0)
        Bnu[:, j] = (2 * h * nu**3 / c**2) / (np.exp(x) - 1.0)

    integrand = R[None, :] * Bnu
    I_nu = np.trapezoid(integrand, R, axis=1)

    Lnu = 4 * np.pi**2 * I_nu
    nuLnu = nu * Lnu

    info = {
        "MBH_solar": MBH_solar,
        "mdot_edd": mdot_edd,
        "Ledd": Ledd,
        "Ldisk": Ldisk,
        "Mdot_cgs": Mdot
    }

    return nu, Lnu, nuLnu, info
```

```python
# ==========================================================
# CALCOLO L_ION
# ==========================================================
def calcola_lion_per_transizioni(nu, Lnu, df_trans, nu_max=None):
    risultati = []

    if nu_max is None:
        nu_max = np.max(nu)

    for _, row in df_trans.iterrows():
        line_name = row["line_name"]
        nu_ion = row["nu_ion"]
        E_ion = row["E_ion"]

        if nu_ion >= nu_max:
            L_ion = 0.0
        else:
            mask = (nu >= nu_ion) & (nu <= nu_max)
            if np.sum(mask) < 2:
                L_ion = 0.0
            else:
                L_ion = np.trapezoid(Lnu[mask], nu[mask])

        risultati.append({
            "line_name": line_name,
            "E_ion": E_ion,
            "nu_ion": nu_ion,
            "L_ion": L_ion
        })

    return pd.DataFrame(risultati)
```

```python
# ==========================================================
# MODELLO: predizione linee da SS
# ==========================================================
def predici_linee_da_modello_ss(logm, logmdot, df_trans, cf_blr=0.1):
    m = 10**logm
    mdot = 10**logmdot

    nu, Lnu, nuLnu, info = costruisci_spettro_ss(
        MBH_solar=m,
        mdot_edd=mdot,
        eta=eta,
        Nr=Nr,
        Nnu=Nnu,
        nu_min=nu_min,
        nu_max=nu_max
    )

    df_lion = calcola_lion_per_transizioni(nu, Lnu, df_trans, nu_max=nu_max)
    df_lion["L_line_mod"] = cf_blr * df_lion["L_ion"]

    return df_lion
```

```python
# ==========================================================
# PREPARAZIONE DATI OSSERVATI
# ==========================================================
def prepara_dati_blazar(label_target, file_linee, file_transizioni):
    df_tab = pd.read_csv(file_linee)
    df_trans = leggi_transizioni(file_transizioni)

    col_name_tab = trova_colonna(df_tab, ["label", "file_name", "name", "nome", "blazar"])
    norm_columns = [col for col in df_tab.columns if col.endswith("_norm")]

    df_all = df_tab.rename(columns={col_name_tab: "label"}).copy()
    df_all["label"] = df_all["label"].astype(str).str.strip()

    label_target_clean = str(label_target).strip()

    df_match = df_all[df_all["label"] == label_target_clean].copy()

    if df_match.empty:
        df_match = df_all[
            df_all["label"].str.contains(label_target_clean, case=False, na=False)
        ].copy()

    if df_match.empty:
        print("Label disponibili:")
        print(df_all["label"].drop_duplicates().sort_values().tolist())
        raise ValueError(f"Nessun blazar trovato per label_target = '{label_target}'")

    if len(df_match) > 1:
        print("Trovati più blazar compatibili:")
        print(df_match["label"].tolist())
        print("Uso il primo della lista.")

    row = df_match.iloc[0]

    df_obs = estrai_linee_osservate(row, norm_columns)

    df_obs = df_obs.merge(
        df_trans[["line_name", "E_ion", "nu_ion"]],
        on="line_name",
        how="left"
    )

    df_obs["sigma_L"] = pd.to_numeric(df_obs["sigma_L"], errors="coerce")
    df_obs.loc[
        df_obs["sigma_L"].isna() | (df_obs["sigma_L"] <= 0),
        "sigma_L"
    ] = REL_ERR * df_obs["L_line_obs"]

    print(f"Blazar selezionato: {row['label']}")
    print(f"Numero linee osservate usate: {len(df_obs)}")

    return df_all, df_trans, df_obs
```

```python
# ==========================================================
# PRIOR, LIKELIHOOD, POSTERIOR
# ==========================================================
def log_prior(theta):
    logm, logmdot = theta

    if not (6.7 < logm < 10.0):             
        return -np.inf

    if not (-2.0 < logmdot < 0.0):
        return -np.inf

    return 0.0


def log_likelihood(theta, df_obs, df_trans, cf_blr=0.1):
    logm, logmdot = theta

    try:
        df_mod = predici_linee_da_modello_ss(
            logm=logm,
            logmdot=logmdot,
            df_trans=df_trans,
            cf_blr=cf_blr
        )
    except Exception:
        return -np.inf

    df_cmp = df_obs.merge(
        df_mod[["line_name", "L_line_mod"]],
        on="line_name",
        how="inner"
    )

    df_cmp = df_cmp[
        df_cmp["L_line_obs"].notna() &
        df_cmp["sigma_L"].notna() &
        df_cmp["L_line_mod"].notna() &
        np.isfinite(df_cmp["L_line_mod"]) &
        (df_cmp["L_line_mod"] >= 0) &
        (df_cmp["sigma_L"] > 0)
    ].copy()

    if len(df_cmp) == 0:
        return -np.inf

    resid = df_cmp["L_line_obs"].values - df_cmp["L_line_mod"].values
    sigma = df_cmp["sigma_L"].values

    if not np.all(np.isfinite(resid)) or not np.all(np.isfinite(sigma)):
        return -np.inf

    return -0.5 * np.sum((resid / sigma)**2 + np.log(2*np.pi*sigma**2))        #formula log-likelihood per error gaussiano con varianza sigma^2


def log_posterior(theta, df_obs, df_trans, cf_blr=0.1):
    lp = log_prior(theta)
    if not np.isfinite(lp):
        return -np.inf

    ll = log_likelihood(theta, df_obs, df_trans, cf_blr=cf_blr)
    if not np.isfinite(ll):
        return -np.inf

    return lp + ll
```

```python
# ==========================================================
# INIZIALIZZAZIONE WALKERS
# ==========================================================
def inizializza_walkers(nwalkers, guess_logm=9.0, guess_logmdot=-1.0,
                        spread_logm=0.05, spread_logmdot=0.05, random_seed=42):
    rng = np.random.default_rng(random_seed)

    pos = []
    while len(pos) < nwalkers:
        trial = np.array([
            rng.normal(guess_logm, spread_logm),
            rng.normal(guess_logmdot, spread_logmdot)
        ])

        if np.isfinite(log_prior(trial)):
            pos.append(trial)

    return np.array(pos)
```

```python
# ==========================================================
# RUN EMCEE
# ==========================================================
def run_emcee_ss(
    df_obs,
    df_trans,
    nwalkers=32,
    nsteps=5000,
    initial_guess=(9.0, -1.0),
    cf_blr=0.1,
    random_seed=42
):
    ndim = 2

    initial_pos = inizializza_walkers(
        nwalkers=nwalkers,
        guess_logm=initial_guess[0],
        guess_logmdot=initial_guess[1],
        spread_logm=0.05,
        spread_logmdot=0.05,
        random_seed=random_seed
    )

    sampler = emcee.EnsembleSampler(
        nwalkers,
        ndim,
        log_posterior,
        args=(df_obs, df_trans, cf_blr)
    )

    sampler.run_mcmc(initial_pos, nsteps, progress=True)

    return sampler
```

```python
# ==========================================================
# RIASSUNTO POSTERIORE
# ==========================================================
def riassumi_posteriore(samples):
    logm = samples[:, 0]
    logmdot = samples[:, 1]

    out = {
        "logm_median": np.median(logm),
        "logm_p16": np.percentile(logm, 16),
        "logm_p84": np.percentile(logm, 84),
        "logmdot_median": np.median(logmdot),
        "logmdot_p16": np.percentile(logmdot, 16),
        "logmdot_p84": np.percentile(logmdot, 84),
    }

    out["m_median"] = 10**out["logm_median"]
    out["mdot_median"] = 10**out["logmdot_median"]

    return out
```

```python
# ==========================================================
# TRACE PLOT
# ==========================================================
def plot_trace_emcee(sampler, burn_in=0, titolo=""):
    chain = sampler.get_chain()   # shape = (nsteps, nwalkers, ndim)
    nsteps, nwalkers, ndim = chain.shape

    fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)

    for i in range(nwalkers):
        axes[0].plot(chain[:, i, 0], alpha=0.35, lw=0.7)
        axes[1].plot(chain[:, i, 1], alpha=0.35, lw=0.7)

    axes[0].axvline(burn_in, color="red", linestyle="--")
    axes[1].axvline(burn_in, color="red", linestyle="--")

    axes[0].set_ylabel(r"$\log_{10}(m)$")
    axes[1].set_ylabel(r"$\log_{10}(\dot m)$")
    axes[1].set_xlabel("Step")

    fig.suptitle(titolo)
    plt.tight_layout()
    plt.show()

# ==========================================================
# CORNER PLOT
# ==========================================================
def plot_corner_emcee(samples, titolo=""):
    fig = corner.corner(
        samples,
        labels=[r"$\log_{10}(m)$", r"$\log_{10}(\dot m)$"],
        quantiles=[0.16, 0.50, 0.84],
        show_titles=True,
        title_kwargs={"fontsize": 12}
    )
    fig.suptitle(titolo, y=1.02)
    plt.show()
```

```python
# ==========================================================
# BEST FIT: confronto linee osservate vs modello
# ==========================================================
def plot_best_fit_linee_ss(df_obs, df_trans, best_logm, best_logmdot, cf_blr=0.1):
    df_mod = predici_linee_da_modello_ss(
        logm=best_logm,
        logmdot=best_logmdot,
        df_trans=df_trans,
        cf_blr=cf_blr
    )

    df_cmp = df_obs.merge(
        df_mod[["line_name", "L_line_mod"]],
        on="line_name",
        how="inner"
    )

    if df_cmp.empty:
        print("Nessuna linea comune tra osservazioni e modello.")
        return

    x = np.arange(len(df_cmp))

    plt.figure(figsize=(9, 5))
    plt.errorbar(
        x,
        df_cmp["L_line_obs"],
        yerr=df_cmp["sigma_L"],
        fmt="o",
        capsize=4,
        label="Osservate"
    )
    plt.scatter(x, df_cmp["L_line_mod"], marker="s", s=70, label="Modello SS")
    plt.yscale("log")
    plt.xticks(x, df_cmp["line_name"], rotation=45)
    plt.ylabel(r"$L_{\rm line}$ [erg s$^{-1}$]")
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
# ==========================================================
# FIT COMPLETO
# ==========================================================
def fit_blazar_mcmc_ss(
    label_target,
    initial_guess=(9.0, -1.0),
    nwalkers=32,
    nsteps=5000,
    burn_in=1000,
    thin=10,
    cf_blr=0.1,
    random_seed=42
):
    df_all, df_trans, df_obs = prepara_dati_blazar(
        label_target=label_target,
        file_linee=file_linee,
        file_transizioni=file_transizioni
    )

    sampler = run_emcee_ss(
        df_obs=df_obs,
        df_trans=df_trans,
        nwalkers=nwalkers,
        nsteps=nsteps,
        initial_guess=initial_guess,
        cf_blr=cf_blr,
        random_seed=random_seed
    )

    chain = sampler.get_chain()
    log_prob = sampler.get_log_prob()

    flat_samples = sampler.get_chain(discard=burn_in, thin=thin, flat=True)

    summary = riassumi_posteriore(flat_samples)

    acc_frac = np.mean(sampler.acceptance_fraction)

    print("=" * 80)
    print(f"Blazar: {label_target}")
    print("Regime: SS")
    print(f"Numero walkers = {nwalkers}")
    print(f"Numero steps   = {nsteps}")
    print(f"Burn-in        = {burn_in}")
    print(f"Thin           = {thin}")
    print(f"Acceptance fraction media = {acc_frac:.3f}")
    print()
    print(f"log10(m)    = {summary['logm_median']:.3f} "
          f"[{summary['logm_p16']:.3f}, {summary['logm_p84']:.3f}]")
    print(f"log10(mdot) = {summary['logmdot_median']:.3f} "
          f"[{summary['logmdot_p16']:.3f}, {summary['logmdot_p84']:.3f}]")
    print(f"m_median    = {summary['m_median']:.3e} Msun")
    print(f"mdot_median = {summary['mdot_median']:.3e}")
    print("=" * 80)

    plot_trace_emcee(
        sampler,
        burn_in=burn_in,
        titolo=f"{label_target} - SS - trace plot"
    )

    plot_corner_emcee(
        flat_samples,
        titolo=f"{label_target} - SS - corner plot"
    )

    plot_best_fit_linee_ss(
        df_obs=df_obs,
        df_trans=df_trans,
        best_logm=summary["logm_median"],
        best_logmdot=summary["logmdot_median"],
        cf_blr=cf_blr
    )

    return {
        "df_obs": df_obs,
        "df_trans": df_trans,
        "sampler": sampler,
        "chain": chain,
        "log_prob": log_prob,
        "flat_samples": flat_samples,
        "summary": summary,
        "acceptance_fraction_mean": acc_frac
    }
```
